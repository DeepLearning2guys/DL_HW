```python
import torch
from torch import nn
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision.transforms import ToTensor
```

```python
# Download training data from open datasets.
training_data = datasets.FashionMNIST(
    root="data",
    train=True,
    download=True,
    transform=ToTensor(),
)

# Download test data from open datasets.
test_data = datasets.FashionMNIST(
    root="data",
    train=False,
    download=True,
    transform=ToTensor(),
)
```

```python
# 每一批大小为64
batch_size = 64

# Create data loaders.
train_dataloader = DataLoader(training_data, batch_size=batch_size)
test_dataloader = DataLoader(test_data, batch_size=batch_size)

for X, y in test_dataloader:
    print(f"Shape of X [N, C, H, W]: {X.shape}")
    print(f"Shape of y: {y.shape} {y.dtype}")
    break
```

```python
import torch
import numpy as np

# (1) 定义常量与变量
# 定义常量
constant = torch.tensor([1, 2, 3], dtype=torch.float32)
print("常量:", constant)

# 定义变量
variable = torch.tensor([4, 5, 6], dtype=torch.float32, requires_grad=True)
print("变量:", variable)
```

```python
# (2) Tensor 与 numpy 的转换
# Tensor 转 numpy
tensor_to_numpy = torch.tensor([7, 8, 9], dtype=torch.float32)
numpy_array = tensor_to_numpy.numpy()
print("Tensor 转 numpy:", numpy_array)

# numpy 转 Tensor
numpy_to_tensor = np.array([10, 11, 12], dtype=np.float32)
tensor_array = torch.from_numpy(numpy_to_tensor)
print("numpy 转 Tensor:", tensor_array)
```

```python
# (3) 数据 reshape 与最大值索引
data = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print("原始数据:", data)

# reshape
reshaped_data = data.reshape(3, 2)
print("reshape 后的数据:", reshaped_data)

# 最大值索引
max_value, max_index = data.max(1)
print("每行最大值:", max_value)
print("每行最大值索引:", max_index)
```

```python
# (4) 随机数据生成
# 生成随机数据
random_data = torch.rand(2, 3)
print("随机数据:", random_data)
```

```python
# (5) 基本算术操作
a = torch.tensor([1, 2, 3], dtype=torch.float32)
b = torch.tensor([4, 5, 6], dtype=torch.float32)

# 加法
add_result = a + b
print("加法结果:", add_result)

# 减法
subtract_result = a - b
print("减法结果:", subtract_result)

# 乘法
multiply_result = a * b
print("乘法结果:", multiply_result)

# 除法
divide_result = a / b
print("除法结果:", divide_result)
```

```python
# (6) 基本卷积操作
# 定义输入数据
input_data = torch.randn(1, 1, 5, 5)  # (batch_size, channels, height, width)
# 定义卷积核
conv_kernel = torch.randn(1, 1, 3, 3)  # (out_channels, in_channels, kernel_height, kernel_width)
# 进行卷积操作
conv_result = torch.nn.functional.conv2d(input_data, conv_kernel, padding=1)
print("卷积操作结果:", conv_result)
```

```python
# (7) GPU 检测与支持
# 检测 GPU 是否可用
if torch.cuda.is_available():
    print("GPU 可用")
    # 将 Tensor 转移到 GPU
    tensor_on_gpu = torch.tensor([1, 2, 3], dtype=torch.float32).cuda()
    print("在 GPU 上的 Tensor:", tensor_on_gpu)
else:
    print("GPU 不可用")
```

```python
# (8) 网络化与 Cat
# 定义两个 Tensor
tensor1 = torch.tensor([1, 2, 3], dtype=torch.float32)
tensor2 = torch.tensor([4, 5, 6], dtype=torch.float32)

# 沿着第 0 维度拼接
cat_result = torch.cat((tensor1, tensor2), dim=0)
print("拼接结果:", cat_result)
```
